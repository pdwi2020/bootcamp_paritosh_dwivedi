## Draft: Scoping paragraph

The Weekly ETF risk monitor addresses a recurring decision problem: a portfolio manager needs a consistent way to judge whether current SPY exposure warrants additional investigation before the weekly risk review. Daily price moves alone do not translate cleanly into the next week's risk. The monitor will use information available at the decision date to forecast annualized realized volatility over the next five trading sessions and classify the period as normal risk or elevated risk. A useful implementation must lower chronological holdout mean absolute error relative to a trailing 20-day volatility baseline and achieve at least 50% elevated-risk recall while reporting balanced accuracy.

The portfolio manager is both the decision owner and primary user. Before the weekly review, the manager receives the volatility forecast, an elevated-risk score, recent return, drawdown, volatility, and volume context, and a plain-language interpretation. An elevated flag prompts review of concentration, liquidity, and available risk controls; a normal flag supports continued monitoring within the existing mandate but is not a claim that the position is safe. The answer is predictive, not causal, and the risk score is a ranking and decision score, not a calibrated event probability.
